```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./datasets/temp.csv')
print(df.head())
```

```python
def vcode_to_time(vcode):
    step = int(vcode[1:])  
    hour = step * 15 // 60
    minute = (step * 15) % 60
    return f"{hour:02}:{minute:02}"

df["HourOfDay"] = df["Time"].apply(vcode_to_time)

plt.figure(figsize=(14, 6))
sns.barplot(x="HourOfDay", y="Volume", data=df)
plt.title("Traffic Volume per Time of Day (15-Min Intervals)")
plt.xlabel("Time of Day")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
