# This Script For Training Model

### Setup Class Model

```python
import numpy as np
import os

from lstm_training import LSTMModel
from gru_training import GRUModel
from cnn_lstm_training import CNNLSTMModel
```

```python
# Load Data
dataset_path = "F:/Swinburne/COS30019/Assignment2B/datasets/dataset.npz"

data = np.load(dataset_path)
print(data.files)

x_train = data['X_train']
y_train = data['y_train']
x_val = data['X_val']
y_val = data['y_val']
x_test = data['X_test']
y_test = data['y_test']

print(x_train.shape)
print(y_train.shape)
print(x_val.shape)
print(y_val.shape)
print(x_test.shape)
print(y_test.shape)
```

### LSTM Training

```python
lstm_model = LSTMModel(input_shape=x_train.shape[1:], epochs=50, use_cosine_decay=True, learning_rate=0.0005, patience=3)
lstm_model.train(x_train, y_train, x_val, y_val, vis=True)
lstm_model.save("F:/Swinburne/COS30019/Assignment2B/checkpoints/lstm/lstm_model.keras")
```

```python
lstm_model_infer = LSTMModel(model_path="F:/Swinburne/COS30019/Assignment2B/checkpoints/lstm/lstm_model.keras")
lstm_model_infer.evaluate(x_test, y_test)
```

### GRU Training

```python
gru_model = GRUModel(input_shape=x_train.shape[1:], epochs=50, use_cosine_decay=True, learning_rate=0.0005, patience=3)
gru_model.train(x_train, y_train, x_val, y_val, vis=True)
gru_model.save("F:/Swinburne/COS30019/Assignment2B/checkpoints/gru/gru_model.keras")
```

```python
gru_model_infer = GRUModel(model_path="F:/Swinburne/COS30019/Assignment2B/checkpoints/gru/gru_model.keras")
gru_model_infer.evaluate(x_test, y_test)
```

### CNNLSTM Training

```python
cnn_lstm_model = CNNLSTMModel(input_shape=x_train.shape[1:], epochs=50, use_cosine_decay=True, learning_rate=0.0005, patience=3)
cnn_lstm_model.train(x_train, y_train, x_val, y_val, vis=True)
cnn_lstm_model.save("F:/Swinburne/COS30019/Assignment2B/checkpoints/cnnlstm/cnn_lstm_model.keras")
```

```python
cnn_lstm_model_infer = CNNLSTMModel(model_path="F:/Swinburne/COS30019/Assignment2B/checkpoints/cnnlstm/cnn_lstm_model.keras")
cnn_lstm_model_infer.evaluate(x_test, y_test)
```
